```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
## Variable independiente
CantidadDeHoras=[120.2,133.5,548.9,254.6,147.2,178,426.5,450,355.7,756.7,351.3]

print("Variable independiente=CantidadDeHoras")

## Variable dependiente
Calificacion=[2,4,9,6,1,3,10,8,7,9,8]

print("Variable Dependiente=Calificacion")
```

```python
df=pd.DataFrame({"CantidadDeHoras":CantidadDeHoras,"Calificacion":Calificacion})
```

```python
df["x*y"]=df["CantidadDeHoras"]*df["Calificacion"]
df["x^2"]=df["CantidadDeHoras"]**2
df["y^2"]=df["Calificacion"]**2
```

```python
print("\n Tabla de datos")
print(df)
```

```python
n=len(df)
```

```python
xSum =df["CantidadDeHoras"].sum()
ySum =df["Calificacion"].sum()
xySum=df["x*y"].sum()
x2Sum=df["x^2"].sum()
y2Sum=df["y^2"].sum()
```

```python
print("\n Sumatoria")

print("Suma X:", xSum)
print("Suma Y:", ySum)
print("Suma XY:", xySum)
print("Suma X^2:", x2Sum)
print("Suma Y^2:", y2Sum)
```

```python
b=((n * xySum) - (xSum * ySum)) / ((n * x2Sum) - (xSum ** 2))

a = ((ySum * x2Sum) - (xySum * xSum)) / ((n * x2Sum) - (xSum ** 2))
```

```python
print("\nRecta de regresion")
print(f" y= {a:.2f} + {b:.2f}x")
```

```python
r = ((n * xySum) - (xSum * ySum)) / np.sqrt(((n * x2Sum) - (xSum ** 2)) * ((n * y2Sum) - (ySum ** 2)))
```

```python
print("\n Correlacion")
print(f"r ={r:.4f}")
```

```python
r2=r**2

print(f"r^2= {r2:.4f}")
```

```python
if r>0:
    print("\n La correlacion es positiva")
elif r<0:
    print("\n La correlacion es negativa")
else:
    print("\n No existe una correlacion")

print(f"El modelo explica aproximadamente el {r2*100:.2f}% de la variabilidad.")
```

```python
x=np.array(CantidadDeHoras)
y=a+b*x
```

```python
plt.scatter(CantidadDeHoras,Calificacion)
plt.plot(x,y)

plt.title("Regresion lineal")
plt.xlabel("CantidadDeHoras")
plt.ylabel("Calificacion")

plt.show()
```

```python
if abs(r)>0.7:
    print("Los datos parecen seguir una tendencia lineal fuerte")
elif abs(r)>0.4:
    print("Los datos parecen seguir una tendencia lineal media ")
else:
    print("Los datos no muestran una tendencia linea clara")
```
