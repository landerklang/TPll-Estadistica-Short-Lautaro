```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
## Variable independiente
Tiempo=[8,14,22,27,33,50,60]

## Variable dependiente
Altura=[17,15,14,12,11,6,4]
```

```python
df=pd.DataFrame({"Tiempo":Tiempo,"Altura":Altura})
```

```python
df["x*y"]=df["Tiempo"]*df["Altura"]
df["x^2"]=df["Tiempo"]**2
df["y^2"]=df["Altura"]**2
```

```python
print("\nTabla de datos\n")
print(df)
```

```python
n=len(df)
```

```python
xSum =df["Tiempo"].sum()
ySum =df["Altura"].sum()
xySum=df["x*y"].sum()
x2Sum=df["x^2"].sum()
y2Sum=df["y^2"].sum()
```

```python
print("\n Sumatoria")

print("Suma X:", xSum)
print("Suma Y:", ySum)
print("Suma XY:", xySum)
print("Suma X^2:", x2Sum)
print("Suma Y^2:", y2Sum)
```

```python
b=((n * xySum) - (xSum * ySum)) / ((n * x2Sum) - (xSum ** 2))

a = ((ySum * x2Sum) - (xySum * xSum)) / ((n * x2Sum) - (xSum ** 2))
```

```python
print("\nRecta de regresion")
print(f" y= {a:.2f} + {b:.2f}x")
```

```python
r = ((n * xySum) - (xSum * ySum)) / np.sqrt(((n * x2Sum) - (xSum ** 2)) * ((n * y2Sum) - (ySum ** 2)))
```

```python
print("\n Correlacion")
print(f"r ={r:.4f}")
```

```python
r2=r**2

print(f"r^2= {r2:.4f}")
```

```python
if r>0:
    print("\n La corrrelacion es positiva")
elif r<0:
    print("\n La correlacion es negativa")
else:
    print("\n No existe una correlacion")

print(f"El modelo explica aproximadamente el {r2*100:.2f}% de la variabilidad.")
```

```python
x=np.array(Tiempo)
y=a+b*x
```

```python
plt.scatter(Tiempo,Altura)
plt.plot(x,y)

plt.title("Regresion lineal")
plt.xlabel("Tiempo")
plt.ylabel("Altura")

plt.show()
```
